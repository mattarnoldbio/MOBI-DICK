```python
import pandas as pd
```

```python
families_virion = [line for line in open('/Users/arno01m/Downloads/VertVirusFamilies.txt')]
families_virion = [line[:-1] for line in families_virion]
```

```python
families_virion

non_vert_families =[
    "mitoviridae", "totiviridae", "partitiviridae", "iflaviridae", "dicistroviridae", 
    "virgaviridae", "picobirnaviridae", "retroviridae", "iridoviridae", 
    "alloherpesviridae", "geminiviridae", "caulimoviridae", "picobirnaviridae", 
    "chuviridae", "betaflexiviridae", "alphasatellitidae", "pospiviroidae", 
    "nanghoshaviridae", "secoviridae", "tospoviridae", "tolecusatellitidae", 
    "amnoonviridae", "bromoviridae", "tombusviridae", "tymoviridae", "tosoviridae", 
    "sunviridae", "carmotetraviridae", "chrysoviridae", "deltaflexiviridae", 
    "marseilleviridae", "nanhypoviridae", "bacilladnaviridae", "blumeviridae", 
    "mesoniviridae", "cremegaviridae", "olifoviridae", "qinviridae", "artoviridae", 
    "gresnaviridae", "endornaviridae", "mymonaviridae", "nenyaviridae", "tristromaviridae"
]
```

```python
vert_families = set(families_virion) - set(non_vert_families)
```

```python
print(list(vert_families))
```

```python
virus_hits = pd.read_csv('/Users/arno01m/Downloads/batch2_final_viral_hits.csv')

vert_virus_hits = virus_hits[virus_hits['family'].isin(vert_families)]
```

```python
print("Total virus hits: ", len(virus_hits))

print("Total vert virus hits: ", len(vert_virus_hits))
```

```python
# convert family column of virus hits to lowercase
virus_hits['family'] = virus_hits['family'].str.lower()

vert_virus_hits = virus_hits[virus_hits['family'].isin(vert_families)]
vert_virus_hits =  vert_virus_hits[vert_virus_hits['length'] >= 300]
```

```python
vert_virus_hits.to_csv('/Users/arno01m/Downloads/vert_virus_hits.csv', index=False)
```
